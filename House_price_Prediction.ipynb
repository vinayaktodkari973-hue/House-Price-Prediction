```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
```

```python
data=pd.read_csv("HousePricePrediction.csv")
data.head()
data = data.dropna(subset=["SalePrice"])
```

```python
#data["MSSubClass"].unique()
```

```python
X=data.drop(columns=["SalePrice","Id"])
Y=data["SalePrice"]
```

```python
#X.head()
```

```python
X=pd.get_dummies(X,columns=["MSZoning"],drop_first=True,dtype=int)
X=pd.get_dummies(X,columns=["LotConfig"],drop_first=True,dtype=int)
X=pd.get_dummies(X,columns=["BldgType"],drop_first=True,dtype=int)
X=pd.get_dummies(X,columns=["Exterior1st"],drop_first=True,dtype=int)
X = pd.get_dummies(X, columns=["MSSubClass"], drop_first=True, dtype=int)
```

```python
#X.head()
```

```python
X["TotalBsmtSF"] = X["TotalBsmtSF"].fillna(X["TotalBsmtSF"].median())
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
model=LinearRegression()
model.fit(X_train,Y_train)
```

```python
Y_predict=model.predict(X_test)
```

```python
print(Y_predict)
```

```python
r2=r2_score(Y_test,Y_predict)
print(r2)
```
